```python
import pandas as pd

file = "eleve_dataset.xlsx"
xls = pd.ExcelFile(file)

print(xls.sheet_names)
```

```python
df_eng = pd.read_excel(xls, "Score engagement (Kpo)")
df_util = pd.read_excel(xls, "Temps total d’utilisation par s")
df_temps = pd.read_excel(xls, "temps_par_matiere")
df_review = pd.read_excel(xls, "review_parent")
```

```python
df_eng.columns = df_eng.columns.str.strip()
df_util.columns = df_util.columns.str.strip()
df_temps.columns = df_temps.columns.str.strip()
df_review.columns = df_review.columns.str.strip()
```

```python
df_final = df_eng[["Candidat", "Section", "Score engagement (Kpo)"]].copy()
```

```python
print(df_eng.columns.tolist())
```

```python
df_final = df_eng[["Candidat", "Section", "Score engagement (Kpo)"]].copy()
```

```python
# nettoyer
df_eng.columns = df_eng.columns.str.strip().str.lower()

# renommer "nom" → "candidat"
df_eng.rename(columns={"nom": "candidat"}, inplace=True)

# créer la base
df_final = df_eng[["candidat", "section", "score engagement (kpo)"]].copy()
```

```python
df_util.columns = df_util.columns.str.strip().str.lower()
df_util.rename(columns={"nom": "candidat"}, inplace=True)

df_temps.columns = df_temps.columns.str.strip().str.lower()
df_temps.rename(columns={"nom": "candidat"}, inplace=True)

df_review.columns = df_review.columns.str.strip().str.lower()
df_review.rename(columns={"nom": "candidat"}, inplace=True)
```

```python
df_util.columns = df_util.columns.str.strip().str.lower()

# transformer "nom" en "candidat"
df_util.rename(columns={"nom": "candidat"}, inplace=True)
```

```python
print(df_util.columns.tolist())
```

```python
df_util_grouped = df_util.groupby("candidat").agg({
    "temps total d’utilisation par semaine (min)": "sum",
    "date de début": "min"
}).reset_index()
```

```python
df_temps.columns = df_temps.columns.str.strip().str.lower()

# renommer nom → candidat
df_temps.rename(columns={"nom": "candidat"}, inplace=True)
```

```python
print(df_temps.columns.tolist())
```

```python
df_temps_grouped = df_temps.groupby("candidat").agg({
    "temps": "sum"
}).reset_index()

df_temps_grouped.rename(columns={"temps": "temps_par_matiere"}, inplace=True)
```

```python
def clean(df):
    df.columns = df.columns.str.strip().str.lower()
    df.rename(columns={"nom": "candidat"}, inplace=True)
    return df

df_eng = clean(df_eng)
df_util = clean(df_util)
df_temps = clean(df_temps)
df_review = clean(df_review)
```

```python
df_review_grouped = df_review.groupby("candidat").agg({
    "satisfait": "sum",
    "non satisfait": "sum",
    "problème temps": "sum",
    "problème connexion": "sum",
    "absence live": "sum",
    "difficultés matières": "sum",
    "travail irrégulier": "sum",
    "motivation": "sum",
    "autre": "sum"
}).reset_index()
```

```python
df_final = df_final.merge(df_util_grouped, on="candidat", how="left")
df_final = df_final.merge(df_temps_grouped, on="candidat", how="left")
df_final = df_final.merge(df_review_grouped, on="candidat", how="left")
```

```python
df_final.fillna("Aucun", inplace=True)
```
